## Q7. Sum of weights

1. Select all the cars from Asia.
2. Keep only the columns `vehicle_weight` and `model_year`.
3. Select the first 7 values.
4. Get the underlying NumPy array — call it `X`.
5. Compute `XTX = X.T @ X`.
6. Invert `XTX`.
7. Create `y = [1100, 1300, 800, 900, 1000, 1100, 1200]`.
8. Compute `w = inv(XTX) @ X.T @ y`.
9. What's the sum of all the elements of `w`?

*Options: 0.0369 / 0.369 / 3.69 / 36.9*

> This is the normal equation for linear regression — covered in Module 2.

```python
# Q7 — steps 1-4: build X
X = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].head(7).to_numpy()
X
```

```python
# Q7 — steps 5-6: XTX and its inverse
XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)
```

```python
# Q7 — steps 7-9: y, w, and the sum
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv @ X.T @ y
w.sum()
```
